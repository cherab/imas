# Render radiation emissivity

Here we try to render a 2D pinhole camera image of a radiation emissivity.

```python
import numpy as np
from raysect.core.math import Vector3D, rotate_basis, rotate_z, translate
from raysect.optical import World
from raysect.optical.observer import PinholeCamera, PowerPipeline2D

from cherab.imas.datasets import iter_solps
from cherab.imas.emitter import load_radiation_emitter
```

## Load radiation emitter

The radiation emitter is loaded from the IMAS dataset containing the radiation IDS.

```python
world = World()
emitter = load_radiation_emitter(
    iter_solps(),
    process_index=[0, 1, 2],
    parent=world,
    interpolator_cache="disk",
)
```

## Render 2D camera view of the radiation

Putting the test camera in position to observe the radiation from the emitter.
The camera will capture a 2D view of the radiation distribution as seen from its position.

```python
R_0 = 15.0  # [m]
R_C = 6.0  # [m]

pipeline = PowerPipeline2D(display_progress=False)
camera = PinholeCamera(
    (256, 256),
    pipelines=[pipeline],
    parent=world,
    transform=translate(0, -R_0, 0)
    * rotate_z(-np.rad2deg(np.arcsin(R_C / R_0)))
    * rotate_basis(Vector3D(0, 1, 0), Vector3D(0, 0, 1)),
)
camera.pixel_samples = 1
```

```python
camera.observe()
```
